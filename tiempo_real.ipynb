```python
import cv2
from ultralytics import YOLO

# 1. Cargar tu modelo entrenado
model = YOLO("best_yolov8n.pt")

# 2. Abrir la cámara (0 = webcam por defecto)
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # 3. Realizar la detección en el fotograma actual
    results = model.predict(frame, conf=0.5)  # Ajusta el conf threshold a tu gusto

    # 4. Tomar el fotograma anotado (con bounding boxes dibujados)
    annotated_frame = results[0].plot()

    # 5. Mostrar el resultado en una ventana
    cv2.imshow("YOLOv8 Real-Time Detection", annotated_frame)

    # 6. Salir con la tecla ESC (27)
    if cv2.waitKey(1) & 0xFF == 27:  
        break

cap.release()
cv2.destroyAllWindows()
```
