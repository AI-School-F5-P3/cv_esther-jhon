```python
import cv2
from ultralytics import YOLO
import os

# Función para seleccionar el modelo
def select_model():
    models = {
        "1": "best_model_25epochs_640px.pt",
        "2": "best_model_50epochs_640px_earlystop.pt"
    }
    print("Seleccione el modelo a utilizar:")
    print("1. Modelo de 25 épocas")
    print("2. Modelo de 50 épocas con early stopping")
    choice = input("Ingrese su elección (1 o 2): ")
    while choice not in models:
        choice = input("Opción inválida. Por favor, ingrese 1 o 2: ")
    return models[choice]

# Función para predicción en imágenes
def predict_image(image_path, model_path):
    """
    Realiza predicciones en una imagen con el modelo YOLOv8.

    Args:
        image_path (str): Ruta a la imagen.
        model_path (str): Ruta al modelo entrenado.
    """
    model = YOLO(model_path)  # Carga el modelo entrenado
    results = model(image_path, save=True)  # Realiza predicciones y guarda los resultados
    print(f"Predicciones guardadas en: {results[0].save_dir}")

# Función para predicción en videos
def predict_video(video_path, model_path):
    """
    Realiza predicciones en un video con el modelo YOLOv8.

    Args:
        video_path (str): Ruta al video.
        model_path (str): Ruta al modelo entrenado.
    """
    model = YOLO(model_path)
    cap = cv2.VideoCapture(video_path)
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        results = model(frame)  # Realiza predicciones en el frame actual
        annotated_frame = results[0].plot()  # Dibuja las anotaciones en el frame
        cv2.imshow("YOLO Detections", annotated_frame)  # Muestra el frame anotado
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break  # Presiona 'q' para salir
    cap.release()
    cv2.destroyAllWindows()

# Punto de entrada principal
if __name__ == "__main__":
    # Seleccionar el modelo
    model_filename = select_model()
    
    # Ruta base al directorio de modelos (ajusta según tu estructura de carpetas)
    BASE_MODEL_DIR = "/content/drive/MyDrive/Proyecto Computer Vision/Entrenamiento yolo"
    
    # Buscar el modelo en las carpetas correspondientes
    if "25epochs" in model_filename:
        MODEL_PATH = os.path.join(BASE_MODEL_DIR, "25_epochs", model_filename)
    elif "50epochs" in model_filename:
        MODEL_PATH = os.path.join(BASE_MODEL_DIR, "50_epochs", model_filename)
    
    if not os.path.exists(MODEL_PATH):
        print(f"Error: No se encontró el modelo en {MODEL_PATH}")
        exit()

    # Rutas de prueba (ajusta según tus datos locales)
    IMAGE_PATH = "path/to/image.jpg"  # Cambia por la ruta de tu imagen
    VIDEO_PATH = "path/to/video.mp4"  # Cambia por la ruta de tu video

    # Predicción en una imagen
    predict_image(IMAGE_PATH, MODEL_PATH)

    # Predicción en un video (descomenta para probar con videos)
    # predict_video(VIDEO_PATH, MODEL_PATH)
```
