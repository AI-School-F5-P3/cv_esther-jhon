```python
import cv2
from ultralytics import YOLO
import os
from datetime import datetime

# Función para seleccionar el modelo
def select_model():
    models = {
        "1": "C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src/models/25epochs/best_model_25epochs_640px.pt",
        "2": "C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src/models/50epochs/best_model_50epochs_640px_earlystop.pt"
    }
    print("Seleccione el modelo a utilizar:")
    print("1. Modelo de 25 épocas")
    print("2. Modelo de 50 épocas con early stopping")
    choice = input("Ingrese su elección (1 o 2): ")
    while choice not in models:
        choice = input("Opción inválida. Por favor, ingrese 1 o 2: ")
    return models[choice]

# Función para predicción en imágenes
def predict_image(image_path, model_path):
    model = YOLO(model_path)
    results = model(image_path, save=True)
    print(f"Predicciones guardadas en: {results[0].save_dir}")

# Función para predicción en videos
def predict_video(video_path, model_path):
    model = YOLO(model_path)
    cap = cv2.VideoCapture(video_path)
    
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = int(cap.get(cv2.CAP_PROP_FPS))
    
    output_dir = os.path.dirname(video_path)
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    output_filename = f"output_video_{timestamp}.mp4"
    output_path = os.path.join(output_dir, output_filename)
    
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        results = model(frame)
        annotated_frame = results[0].plot()
        out.write(annotated_frame)
        cv2.imshow("YOLO Detections", annotated_frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    out.release()
    cv2.destroyAllWindows()
    
    print(f"Video procesado guardado en: {output_path}")

# Punto de entrada principal
if __name__ == "__main__":
    model_filename = select_model()
    
    BASE_MODEL_DIR = "/content/drive/MyDrive/Proyecto Computer Vision/Entrenamiento yolo"
    
    if "25epochs" in model_filename:
        MODEL_PATH = os.path.join(BASE_MODEL_DIR, "25_epochs", model_filename)
    elif "50epochs" in model_filename:
        MODEL_PATH = os.path.join(BASE_MODEL_DIR, "50_epochs", model_filename)
    
    if not os.path.exists(MODEL_PATH):
        print(f"Error: No se encontró el modelo en {MODEL_PATH}")
        exit()

    IMAGE_PATH = "C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src/pepsi-lata-33cl-5sentidos.png"
    VIDEO_PATH = "C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src/All the Best Moments are Better With Pepsi.mp4"

    # Predicción en una imagen
    # predict_image(IMAGE_PATH, MODEL_PATH)

    # Predicción en un video
    predict_video(VIDEO_PATH, MODEL_PATH)
```
