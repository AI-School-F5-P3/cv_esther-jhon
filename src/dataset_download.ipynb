```python
from roboflow import Roboflow
```

```python
def download_dataset(api_key, project_name, version):
    """
    Descarga el dataset desde Roboflow y lo configura para YOLO.
    
    Args:
        api_key (str): Clave API de Roboflow.
        project_name (str): Nombre del proyecto en Roboflow.
        version (int): Versión del dataset a descargar.

    Returns:
        str: Ruta al archivo dataset.yaml.
    """
    rf = Roboflow(api_key=api_key)
    project = rf.workspace().project(project_name)
    dataset = project.version(version).download("yolov8")
    return dataset.location  # Ruta del dataset descargado

if __name__ == "__main__":
    API_KEY = 'aeUJ4JbBUUjZzr2QVzk9'  # Cambia esto por tu API Key
    PROJECT_NAME = 'bebidas_web-vcnve'
    VERSION = 1  # Cambia esto por la versión del dataset
    
    dataset_path = download_dataset(API_KEY, PROJECT_NAME, VERSION)
    print(f"Dataset descargado en: {dataset_path}")
```
