# Entrenamiento 25 épocas

```python
# Importamos las librerías necesarias
from ultralytics import YOLO
import os
from datetime import datetime

# Configuración de rutas y parámetros
DATA_PATH = "/content/drive/MyDrive/Proyecto Computer Vision/Bebidas_web-1/data.yaml" #modificar ruta
SAVE_DIR = "/content/drive/MyDrive/Proyecto Computer Vision/Entrenamiento yolo"#modificar ruta

# Parámetros de entrenamiento
EPOCHS = 25
IMG_SIZE = 640

# Creamos una carpeta con un nombre descriptivo para guardar los resultados
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
results_dir = os.path.join(SAVE_DIR, f"YOLOv8_{EPOCHS}epochs_{IMG_SIZE}px_{timestamp}")  # Nombre intuitivo
os.makedirs(results_dir, exist_ok=True)

# Cargamos el modelo preentrenado YOLOv8 Small
model = YOLO("/content/drive/MyDrive/Proyecto Computer Vision/yolov8s.pt")  # #modificar ruta

# Entrenamos el modelo con los parámetros especificados
results = model.train(
    data=DATA_PATH,
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    project=results_dir,  # Los resultados del entrenamiento se guardan aquí
    name="model",         # Subcarpeta específica para este modelo
    save=True             # Guardar el mejor modelo automáticamente
)

print(f"Entrenamiento completado. Resultados guardados en: {results_dir}")
```

```python
# Importamos las librerías necesarias
import shutil
import os

# Intentamos guardar el modelo manualmente
manual_save_path = os.path.join(results_dir, 'best_manual.pt')
model.save(manual_save_path)
print(f"Intento de guardar el modelo manualmente en: {manual_save_path}")

# Verificamos si el archivo se ha creado correctamente
if os.path.exists(manual_save_path):
    print(f"El archivo {manual_save_path} se ha creado correctamente.")
    print(f"Tamaño del archivo: {os.path.getsize(manual_save_path) / (1024 * 1024):.2f} MB")
    
    # Copiamos el archivo manual a la ubicación final con nombre descriptivo
    final_model_path = os.path.join(results_dir, f"best_model_{EPOCHS}epochs_{IMG_SIZE}px.pt")
    shutil.copy(manual_save_path, final_model_path)
    print(f"El mejor modelo se ha copiado a: {final_model_path}")
else:
    print(f"No se pudo crear el archivo {manual_save_path} manualmente.")

# Guardamos las métricas en un archivo de texto con un nombre descriptivo
metrics_file = os.path.join(results_dir, f"training_metrics_{EPOCHS}epochs_{IMG_SIZE}px.txt")
if hasattr(results, "results_dict"):
    with open(metrics_file, "w", encoding="utf-8") as f:
        f.write("Resumen del entrenamiento:\n")
        f.write(f"Épocas: {EPOCHS}\n")
        f.write(f"Tamaño de imagen: {IMG_SIZE}\n")
        f.write("Resultados de métricas:\n")
        for key, value in results.results_dict.items():
            f.write(f"{key}: {value}\n")
    print(f"Métricas guardadas en: {metrics_file}")
else:
    print("Error: No se encontraron métricas en el objeto 'results'.")
```

# Entrenamiento 50 épocas con early stopping

```python
# Importamos las librerías necesarias
from ultralytics import YOLO
import os
from datetime import datetime
import shutil

# Configuración de rutas y parámetros
DATA_PATH = "/content/drive/MyDrive/Proyecto Computer Vision/Bebidas_web-1/data.yaml"#Cambiar ruta
SAVE_DIR = "/content/drive/MyDrive/Proyecto Computer Vision/Entrenamiento yolo"#Cambiar ruta

# Parámetros de entrenamiento
EPOCHS = 50
IMG_SIZE = 640
EARLY_STOPPING_PATIENCE = 5

# Creamos una carpeta con un nombre descriptivo para guardar los resultados
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
results_dir = os.path.join(SAVE_DIR, "50_epochs", f"YOLOv8_{EPOCHS}epochs_{IMG_SIZE}px_{timestamp}_earlystop")
os.makedirs(results_dir, exist_ok=True)

# Cargamos el modelo preentrenado YOLOv8 Small
model = YOLO("/content/drive/MyDrive/Proyecto Computer Vision/yolov8s.pt")#Cambiar ruta

# Entrenamos el modelo con los parámetros especificados, añadiendo early stopping
results = model.train(
    data=DATA_PATH,
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    project=results_dir,
    name="model_50ep_earlystop",
    save=True,
    patience=EARLY_STOPPING_PATIENCE
)

print(f"Entrenamiento completado. Resultados guardados en: {results_dir}")
```

```python
# Intentamos guardar el modelo manualmente
manual_save_path = os.path.join(results_dir, 'best_manual_50ep_earlystop.pt')
model.save(manual_save_path)
print(f"Intento de guardar el modelo manualmente en: {manual_save_path}")

# Verificamos si el archivo se ha creado correctamente
if os.path.exists(manual_save_path):
    print(f"El archivo {manual_save_path} se ha creado correctamente.")
    print(f"Tamaño del archivo: {os.path.getsize(manual_save_path) / (1024 * 1024):.2f} MB")
    
    # Copiamos el archivo manual a la ubicación final con nombre descriptivo
    final_model_path = os.path.join(results_dir, f"best_model_{EPOCHS}epochs_{IMG_SIZE}px_earlystop.pt")
    shutil.copy(manual_save_path, final_model_path)
    print(f"El mejor modelo se ha copiado a: {final_model_path}")
else:
    print(f"No se pudo crear el archivo {manual_save_path} manualmente.")

# Guardamos las métricas en un archivo de texto con un nombre descriptivo
metrics_file = os.path.join(results_dir, f"training_metrics_{EPOCHS}epochs_{IMG_SIZE}px_earlystop.txt")
if hasattr(results, "results_dict"):
    with open(metrics_file, "w", encoding="utf-8") as f:
        f.write("Resumen del entrenamiento:\n")
        f.write(f"Épocas: {EPOCHS}\n")
        f.write(f"Tamaño de imagen: {IMG_SIZE}\n")
        f.write(f"Early Stopping Patience: {EARLY_STOPPING_PATIENCE}\n")
        f.write("Resultados de métricas:\n")
        for key, value in results.results_dict.items():
            f.write(f"{key}: {value}\n")
    print(f"Métricas guardadas en: {metrics_file}")
else:
    print("Error: No se encontraron métricas en el objeto 'results'.")
```
