# Entrenamiento 25 épocas

```python
# Importamos las librerías necesarias
from ultralytics import YOLO
import os
from datetime import datetime
import shutil

# Configuración de rutas y parámetros
DATA_PATH = "C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src/Bebidas_web-1/data.yaml"
SAVE_DIR = "C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src"

# Parámetros de entrenamiento
EPOCHS = 25
IMG_SIZE = 640

# Creamos una carpeta con timestamp para guardar los resultados
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
results_dir = os.path.join(SAVE_DIR, f"train_{timestamp}")
os.makedirs(results_dir, exist_ok=True)

# Cargamos el modelo preentrenado YOLOv8 Small
model = YOLO("C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src/yolov8s.pt")

# Entrenamos el modelo con los parámetros especificados
results = model.train(
    data=DATA_PATH,
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    project=results_dir,
    name="model",
    save=True  # Guardamos el mejor modelo
)

# Guardamos las métricas en un archivo de texto con codificación UTF-8
metrics_file = os.path.join(results_dir, "training_metrics.txt")
with open(metrics_file, "w", encoding="utf-8") as f:
    f.write("Resumen del entrenamiento:\n")
    f.write(f"Épocas: {EPOCHS}\n")
    f.write(f"Tamaño de imagen: {IMG_SIZE}\n")
    f.write("Resultados de métricas:\n")
    for key, value in results.results_dict.items():
        f.write(f"{key}: {value}\n")

# Copiamos el mejor modelo a una ubicación fácil de encontrar
best_model_path = os.path.join(results_dir, "model", "weights", "best.pt")
final_model_path = os.path.join(results_dir, "best_model.pt")
shutil.copy(best_model_path, final_model_path)

print(f"\nEntrenamiento completado!")
print(f"El mejor modelo se ha guardado en: {final_model_path}")
print("\nPara usar este modelo en Google Colab:")
print("1. En el panel izquierdo de archivos, navega hasta la carpeta:", results_dir)
print("2. Encuentra el archivo 'best_model.pt'")
print("3. Haz clic derecho sobre él y selecciona 'Descargar'")
```

# Entrenamiento 50 épocas con eraly stopping

```python
# Importamos las librerías necesarias
from ultralytics import YOLO
import os
from datetime import datetime
import shutil

# Asegúrate de tener instalada la librería torch
# Puedes instalarla ejecutando: pip install torch

# Configuración de rutas y parámetros
DATA_PATH = "C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src/Bebidas_web-1/data.yaml"
SAVE_DIR = "C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src"

# Parámetros de entrenamiento
EPOCHS = 50  # Cambiado a 50 épocas
IMG_SIZE = 640
EARLY_STOPPING_PATIENCE = 5  # Detener si no mejora tras 5 épocas

# Creamos una carpeta con timestamp para guardar los resultados
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
results_dir = os.path.join(SAVE_DIR, f"train_50ep_{timestamp}")
os.makedirs(results_dir, exist_ok=True)

# Configuración de Early Stopping
callbacks = [
    {
        "name": "EarlyStopping",  # Nombre del callback
        "patience": EARLY_STOPPING_PATIENCE  # Número de épocas sin mejora
    }
]

# Cargamos el modelo preentrenado YOLOv8 Small
model = YOLO("C:/Users/Administrator/Desktop/Proyecto_Proyecto CV - Detección de Objetos/cv_esther-jhon/src/yolov8s.pt")

# Entrenamos el modelo con los parámetros especificados
results = model.train(
    data=DATA_PATH,
    epochs=EPOCHS,
    imgsz=IMG_SIZE,
    project=results_dir,
    name="model_50ep",
    save=True,  # Guardamos el mejor modelo
    callbacks=callbacks  # Añadimos los callbacks para early stopping
)

# Guardamos las métricas en un archivo de texto con codificación UTF-8
metrics_file = os.path.join(results_dir, "training_metrics_50ep.txt")
with open(metrics_file, "w", encoding="utf-8") as f:
    f.write("Resumen del entrenamiento:\n")
    f.write(f"Épocas: {EPOCHS}\n")
    f.write(f"Tamaño de imagen: {IMG_SIZE}\n")
    f.write(f"Early Stopping - Patience: {EARLY_STOPPING_PATIENCE}\n")
    f.write("Resultados de métricas:\n")
    for key, value in results.results_dict.items():
        f.write(f"{key}: {value}\n")

# Copiamos el mejor modelo a una ubicación fácil de encontrar
best_model_path = os.path.join(results_dir, "model_50ep", "weights", "best.pt")
final_model_path = os.path.join(results_dir, "best_model_50ep.pt")
shutil.copy(best_model_path, final_model_path)

print(f"\nEntrenamiento completado con 50 épocas y Early Stopping!")
print(f"El mejor modelo se ha guardado en: {final_model_path}")
print("\nPara usar este modelo en Google Colab:")
print("1. En el panel izquierdo de archivos, navega hasta la carpeta:", results_dir)
print("2. Encuentra el archivo 'best_model_50ep.pt'")
print("3. Haz clic derecho sobre él y selecciona 'Descargar'")
```
